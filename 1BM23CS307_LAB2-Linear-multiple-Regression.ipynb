```python
import pandas as pd
import numpy as np

# Load the dataset into a pandas DataFrame
df = pd.read_csv('/content/canada_per_capita_income.csv')

# Extract 'year' and convert to NumPy array
years_np = df['year'].values

# Extract 'per capita income (US$)' and convert to NumPy array
income_np = df['per capita income (US$)'].values

# Print the first 5 elements of both arrays
print("First 5 years (NumPy array):", years_np[:5])
print("First 5 per capita incomes (NumPy array):", income_np[:5])
```

```python
X_with_bias = np.c_[np.ones(years_np.shape[0]), years_np]

print("Shape of X with bias:", X_with_bias.shape)
print("First 5 rows of X with bias:\n", X_with_bias[:5])
```

```python
X_T = X_with_bias.T
X_T_X = X_T.dot(X_with_bias)
X_T_X_inv = np.linalg.inv(X_T_X)
X_T_y = X_T.dot(income_np)
theta = X_T_X_inv.dot(X_T_y)

intercept = theta[0]
slope = theta[1]

print(f"Calculated Intercept: {intercept}")
print(f"Calculated Slope: {slope}")
```

```python
year_2020_np = np.array([2020])

# Add a bias term to the year_2020_np to match the format of X_with_bias
year_2020_with_bias = np.c_[np.ones(year_2020_np.shape[0]), year_2020_np]

# Calculate the predicted per capita income for 2020 using the calculated intercept and slope
predicted_income_2020_np = intercept + slope * year_2020_np[0]

print(f"Predicted per capita income for 2020 (NumPy): {predicted_income_2020_np:,.2f} US$")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

# 1. Scatter plot of historical data
plt.scatter(years_np, income_np, color='blue', label='Historical Data')

# 2. Calculate and plot the regression line
regression_line_predictions = intercept + slope * years_np
plt.plot(years_np, regression_line_predictions, color='red', label='Regression Line')

# 3. Highlight the predicted income for 2020
plt.scatter(year_2020_np[0], predicted_income_2020_np, color='green', s=100, zorder=5, label='2020 Prediction')

# 4. Add labels and title
plt.xlabel('Year')
plt.ylabel('Per Capita Income (US$)')
plt.title('Canada Per Capita Income Prediction (NumPy Model)')

# 5. Add legend
plt.legend()

# 6. Show the plot
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import numpy as np

# Load the dataset into a pandas DataFrame
df_salary = pd.read_csv('/content/salary.csv')

# Handle missing values in 'YearsExperience'
df_salary.dropna(subset=['YearsExperience'], inplace=True)

# Extract 'YearsExperience' and convert to NumPy array
years_exp_np = df_salary['YearsExperience'].values

# Extract 'Salary' and convert to NumPy array
salary_np = df_salary['Salary'].values

# Add a bias term to the years_exp_np array
X_salary_with_bias = np.c_[np.ones(years_exp_np.shape[0]), years_exp_np]

print("Shape of X_salary_with_bias:", X_salary_with_bias.shape)
print("First 5 rows of X_salary_with_bias:\n", X_salary_with_bias[:5])
print("Shape of salary_np:", salary_np.shape)
print("First 5 elements of salary_np:", salary_np[:5])
```

```python
# Calculate coefficients using the normal equation
X_T_salary = X_salary_with_bias.T
X_T_X_salary = X_T_salary.dot(X_salary_with_bias)
X_T_X_inv_salary = np.linalg.inv(X_T_X_salary)
X_T_y_salary = X_T_salary.dot(salary_np)
theta_salary = X_T_X_inv_salary.dot(X_T_y_salary)

intercept_salary = theta_salary[0]
slope_salary = theta_salary[1]

print(f"Calculated Intercept (NumPy): {intercept_salary:.2f}")
print(f"Calculated Slope (NumPy): {slope_salary:.2f}")
```

```python
# Predict salary for 12 years of experience
years_to_predict = np.array([12])
predicted_salary_np_only = intercept_salary + slope_salary * years_to_predict[0]

print(f"Predicted Salary for {years_to_predict[0]} Years Experience (NumPy): ${predicted_salary_np_only:,.2f}")
```

```python
import pandas as pd
df_hiring = pd.read_csv('/content/hiring.csv')
df_hiring.head()
```

```python
df_hiring['experience'] = df_hiring['experience'].astype(float)
median_experience = df_hiring['experience'].median()
df_hiring['experience'] = df_hiring['experience'].fillna(median_experience)

# Select feature columns and target column, explicitly creating a copy to avoid SettingWithCopyWarning
X_hiring = df_hiring[['experience', 'test_score(out of 10)', 'interview_score(out of 10)']].copy()
y_hiring = df_hiring['salary($)'].copy()

# Convert test_score and interview_score to numeric, handle NaNs
X_hiring['test_score(out of 10)'] = pd.to_numeric(X_hiring['test_score(out of 10)'], errors='coerce')
median_test_score = X_hiring['test_score(out of 10)'].median()
X_hiring['test_score(out of 10)'] = X_hiring['test_score(out of 10)'].fillna(median_test_score)

X_hiring['interview_score(out of 10)'] = pd.to_numeric(X_hiring['interview_score(out of 10)'], errors='coerce')
median_interview_score = X_hiring['interview_score(out of 10)'].median()
X_hiring['interview_score(out of 10)'] = X_hiring['interview_score(out of 10)'].fillna(median_interview_score)

# Convert to NumPy arrays
X_hiring_np = X_hiring.values
y_hiring_np = y_hiring.values

# Add a bias term (column of ones) to X_hiring_np
X_hiring_biased = np.c_[np.ones(X_hiring_np.shape[0]), X_hiring_np]

# Calculate coefficients using the normal equation
X_T_X = X_hiring_biased.T @ X_hiring_biased
X_T_X_inv = np.linalg.inv(X_T_X)
X_T_y = X_hiring_biased.T @ y_hiring_np
theta_hiring = X_T_X_inv @ X_T_y

print("Shape of X_hiring_biased:", X_hiring_biased.shape)
print("First 5 rows of X_hiring_biased:\n", X_hiring_biased[:5])
print("Calculated Coefficients (theta_hiring):", theta_hiring)
```

```python
import numpy as np

# Features for the first candidate: 2 years experience, 9 test score, 6 interview score
candidate1_features = np.array([2, 9, 6])

# Add a bias term to the candidate's features
candidate1_biased = np.insert(candidate1_features, 0, 1)

# Predict the salary for the first candidate
predicted_salary_candidate1 = candidate1_biased @ theta_hiring

print(f"Predicted Salary for Candidate 1 (2 yrs exp, 9 test, 6 interview): ${predicted_salary_candidate1:,.2f}")
```

```python
import numpy as np

# Features for the second candidate: 12 years experience, 10 test score, 10 interview score
candidate2_features = np.array([12, 10, 10])

# Add a bias term to the candidate's features
candidate2_biased = np.insert(candidate2_features, 0, 1)

# Predict the salary for the second candidate
predicted_salary_candidate2 = candidate2_biased @ theta_hiring

print(f"Predicted Salary for Candidate 2 (12 yrs exp, 10 test, 10 interview): ${predicted_salary_candidate2:,.2f}")
```

```python
import pandas as pd
df_companies = pd.read_csv('/content/1000_Companies.csv')
df_companies.head()
```

```python
X_companies = df_companies[['R&D Spend', 'Administration', 'Marketing Spend', 'State']]
y_companies = df_companies['Profit']

print("Features (X_companies) head:\n", X_companies.head())
print("\nTarget (y_companies) head:\n", y_companies.head())
```

```python
X_companies_encoded = pd.get_dummies(X_companies, columns=['State'], drop_first=True)

print("Features (X_companies_encoded) head after one-hot encoding:\n", X_companies_encoded.head())
print("Shape of X_companies_encoded:", X_companies_encoded.shape)
```

```python
from sklearn.linear_model import LinearRegression

# Instantiate a Linear Regression model
model_companies = LinearRegression()

print("LinearRegression model instantiated successfully.")
```

```python
model_companies.fit(X_companies_encoded, y_companies)

print("Linear Regression model for companies trained successfully.")
```

```python
new_company_data = pd.DataFrame({
    'R&D Spend': [160000],
    'Administration': [130000],
    'Marketing Spend': [300000],
    'State': ['California']
})

print("New company data DataFrame:")
print(new_company_data)
```

```python
new_company_data_encoded = pd.get_dummies(new_company_data, columns=['State'], drop_first=True)

# Align columns of new data with the training data's encoded columns
# This ensures that if a state is missing in the new data but present in training, or vice versa,
# the column structure remains consistent.
missing_cols = set(X_companies_encoded.columns) - set(new_company_data_encoded.columns)
for c in missing_cols:
    new_company_data_encoded[c] = 0
new_company_data_encoded = new_company_data_encoded[X_companies_encoded.columns]

print("New company data after one-hot encoding and alignment:")
print(new_company_data_encoded)
```

```python
predicted_profit = model_companies.predict(new_company_data_encoded)

print(f"Predicted Profit for the new company: ${predicted_profit[0]:,.2f}")
```
