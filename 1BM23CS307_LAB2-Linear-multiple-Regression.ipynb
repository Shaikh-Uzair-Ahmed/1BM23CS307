# Task
Using the "canada_per_capita_income.csv" file, predict Canada's per capita income for the year 2020 using a linear regression model, visualize the prediction alongside historical data and the regression line, and interpret the results.

## Load Data

### Subtask:
Load the 'canada_per_capita_income.csv' dataset into a pandas DataFrame.


**Reasoning**:
To load the dataset and display its initial rows, I need to import the pandas library, then use `read_csv` to load the specified file into a DataFrame, and finally apply the `.head()` method to show the first few entries.



```python
import pandas as pd
df = pd.read_csv('/content/canada_per_capita_income.csv')
df.head()
```

## Prepare Data for Modeling

### Subtask:
Extract the 'year' as the feature (X) and 'per capita income (US$)' as the target (y) for the regression model.


**Reasoning**:
To prepare the data for the regression model, I need to separate the feature (year) and the target (per capita income). The 'year' column needs to be reshaped into a 2D array as required by scikit-learn models.



```python
X = df[['year']]
y = df['per capita income (US$)']

print("Shape of X:", X.shape)
print("Shape of y:", y.shape)
print("First 5 rows of X:\n", X.head())
print("First 5 rows of y:\n", y.head())
```

## Train Linear Regression Model

### Subtask:
Train a linear regression model using the prepared historical data. This will involve using a library like scikit-learn to fit a linear model to the 'year' and 'per capita income' data.


**Reasoning**:
To train a linear regression model, I need to import the `LinearRegression` class, instantiate it, and then fit it to the prepared data `X` and `y`. Finally, I will print the coefficient and intercept to inspect the model's parameters.



```python
from sklearn.linear_model import LinearRegression

# Create a Linear Regression model object
reg = LinearRegression()

# Train the model using the training sets X and y
reg.fit(X, y)

# Print the coefficient and intercept
print("Coefficient:", reg.coef_[0])
print("Intercept:", reg.intercept_)
```

## Predict 2020 Income

### Subtask:
Use the trained linear regression model to predict the per capita income for the year 2020.


**Reasoning**:
To predict the per capita income for the year 2020, I need to create a 2D array for the year 2020 and then use the `predict` method of the trained linear regression model.



```python
year_2020 = [[2020]]
predicted_income_2020 = reg.predict(year_2020)
print(f"Predicted per capita income for 2020: {predicted_income_2020[0]:,.2f} US$")
```

**Reasoning**:
The previous code produced a `UserWarning` because the input `year_2020` was not provided as a DataFrame with feature names, while the `LinearRegression` model was fitted with a DataFrame. To resolve this, I will create `year_2020` as a pandas DataFrame with the 'year' column name before making the prediction.



```python
import pandas as pd
year_2020_df = pd.DataFrame({'year': [2020]})
predicted_income_2020 = reg.predict(year_2020_df)
print(f"Predicted per capita income for 2020: {predicted_income_2020[0]:,.2f} US$")
```

## Visualize Prediction

### Subtask:
Create a scatter plot of the historical per capita income versus year, overlay the regression line, and highlight the predicted income for 2020. Ensure all elements are clearly labeled with legends.


**Reasoning**:
To visualize the historical data, the regression line, and the predicted 2020 income, I need to use `matplotlib.pyplot` to create a scatter plot, plot the regression line, and highlight the predicted point with appropriate labels and a legend.



```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

# 1. Scatter plot of historical data
plt.scatter(X, y, color='blue', label='Historical Data')

# 2. Calculate and plot the regression line
plt.plot(X, reg.predict(X), color='red', label='Regression Line')

# 3. Highlight the predicted income for 2020
plt.scatter(year_2020_df, predicted_income_2020, color='green', s=100, zorder=5, label='2020 Prediction')

# 4. Add labels and title
plt.xlabel('Year')
plt.ylabel('Per Capita Income (US$)')
plt.title('Canada Per Capita Income Prediction')

# 5. Add legend
plt.legend()

# 6. Show the plot
plt.grid(True)
plt.show()
```

## Final Task

### Subtask:
Present the predicted per capita income for Canada in 2020 and provide a brief interpretation of the results.

# Task
Load the "canada_per_capita_income.csv" dataset into a pandas DataFrame, convert the 'year' and 'per capita income (US$)' columns into NumPy arrays for features and target respectively, and add a constant (bias) term to the 'year' array to prepare it for linear regression using the normal equation.

## Load Data

### Subtask:
Load the 'canada_per_capita_income.csv' dataset into a pandas DataFrame and then convert the relevant columns to NumPy arrays.


**Reasoning**:
I need to load the specified CSV file into a pandas DataFrame and then convert the 'year' and 'per capita income (US$)' columns into NumPy arrays as requested by the subtask. After that, I will print the first 5 elements of each NumPy array to verify the conversion.



```python
import pandas as pd
import numpy as np

# Load the dataset into a pandas DataFrame
df = pd.read_csv('/content/canada_per_capita_income.csv')

# Extract 'year' and convert to NumPy array
years_np = df['year'].values

# Extract 'per capita income (US$)' and convert to NumPy array
income_np = df['per capita income (US$)'].values

# Print the first 5 elements of both arrays
print("First 5 years (NumPy array):", years_np[:5])
print("First 5 per capita incomes (NumPy array):", income_np[:5])
```

**Reasoning**:
The subtask requires adding a constant (bias) term to the 'year' array. To do this, I will reshape `years_np` to a 2D array and then use `np.c_` to concatenate a column of ones to it.



```python
X_with_bias = np.c_[np.ones(years_np.shape[0]), years_np]

print("Shape of X with bias:", X_with_bias.shape)
print("First 5 rows of X with bias:\n", X_with_bias[:5])
```

## Implement Linear Regression with NumPy

### Subtask:
Calculate the coefficients (slope and intercept) for the linear regression model using NumPy's capabilities (e.g., matrix multiplication and inverse) to solve the normal equation: theta = (X^T * X)^-1 * X^T * y.


**Reasoning**:
To calculate the coefficients (theta) using the normal equation, I will follow the specified steps: calculate the transpose of X_with_bias, perform matrix multiplications, compute the inverse, and finally dot product to get theta. Then, I will print the intercept and slope from the resulting theta.



```python
X_T = X_with_bias.T
X_T_X = X_T.dot(X_with_bias)
X_T_X_inv = np.linalg.inv(X_T_X)
X_T_y = X_T.dot(income_np)
theta = X_T_X_inv.dot(X_T_y)

intercept = theta[0]
slope = theta[1]

print(f"Calculated Intercept: {intercept}")
print(f"Calculated Slope: {slope}")
```

## Predict 2020 Income with NumPy

### Subtask:
Use the coefficients calculated with NumPy to predict the per capita income for the year 2020. Ensure the 2020 input is formatted correctly with a bias term.


**Reasoning**:
I need to define the target year, format it with a bias term as instructed, and then use the previously calculated NumPy coefficients (intercept and slope) to predict the per capita income for 2020. Finally, I will print the result.



```python
year_2020_np = np.array([2020])

# Add a bias term to the year_2020_np to match the format of X_with_bias
year_2020_with_bias = np.c_[np.ones(year_2020_np.shape[0]), year_2020_np]

# Calculate the predicted per capita income for 2020 using the calculated intercept and slope
predicted_income_2020_np = intercept + slope * year_2020_np[0]

print(f"Predicted per capita income for 2020 (NumPy): {predicted_income_2020_np:,.2f} US$")
```

## Visualize Prediction

### Subtask:
Create a scatter plot of the historical per capita income versus year, overlay the regression line calculated using the NumPy coefficients, and highlight the predicted income for 2020. Ensure all elements are clearly labeled with legends.


**Reasoning**:
To visualize the historical data, the regression line calculated using NumPy coefficients, and the predicted 2020 income, I need to use `matplotlib.pyplot` to create a scatter plot, plot the regression line, and highlight the predicted point with appropriate labels and a legend, as per the subtask instructions.



```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

# 1. Scatter plot of historical data
plt.scatter(years_np, income_np, color='blue', label='Historical Data')

# 2. Calculate and plot the regression line
regression_line_predictions = intercept + slope * years_np
plt.plot(years_np, regression_line_predictions, color='red', label='Regression Line')

# 3. Highlight the predicted income for 2020
plt.scatter(year_2020_np[0], predicted_income_2020_np, color='green', s=100, zorder=5, label='2020 Prediction')

# 4. Add labels and title
plt.xlabel('Year')
plt.ylabel('Per Capita Income (US$)')
plt.title('Canada Per Capita Income Prediction (NumPy Model)')

# 5. Add legend
plt.legend()

# 6. Show the plot
plt.grid(True)
plt.show()
```
